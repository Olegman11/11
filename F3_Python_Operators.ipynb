```python
MyFruit = ['apple', 'orange', 'banana', 'kiwi']
SubFruit = MyFruit[2:3]
print(SubFruit)

MyFruit.append('Carrot') # it adds 'Carrot' in the list
print(MyFruit)
print(len(MyFruit))
```

```python
MyFruit.remove('apple') # it removes 'Carrot' in the list
print(MyFruit)
print(len(MyFruit))
```

```python
MyFruit.pop(1) # 'pop' - it takes out the item from the list
print(MyFruit)
print(len(MyFruit))
```

```python
fr = ['apple', 'banana', 'grapes', 'orange', 'banana', 'kiwi']
fr.remove('orange')  # remove from the list
fr.remove('grapes')  # append to the list

print(fr)
print(len(fr))
fr.append('orange') # append to the list
fr.append('grapes')  # append to the list

print(fr)
print(len(fr))
```
