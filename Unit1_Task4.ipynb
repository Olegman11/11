**Task 4 - data report and practical task (LO3)**
*Note: Although this is a very small dataset and cleaning it up in Excel or Google Sheets is not particularly difficult and would take just five minutes (deleting unnecessary rows and columns, renaming cells and changing their data types), I decided to do it in Python to improve my Python skills.*

```python
# turning on pandas lab
import pandas as pd
# turning on numpy
import numpy as np
```

```python
# uploading datasets files
students_data = pd.read_csv('/datafile1_students_information.csv')
jobs_data = pd.read_csv('/datafile2_jobs.csv', encoding='cp1252')
course_history = pd.read_csv('/datafile3_course_data (1).csv')
```

```python
## Make a copy of my datasets in case I lose the original dataset
students_data2 = students_data.copy()
jobs_data2 = jobs_data.copy()
course_history2 = course_history.copy()
```

Clearing the “Students information” table

```python
# info about students data
students_data2.info()
print(students_data2.columns)
```

```python
# found an empty column 3 Unnamed: 3 0 non-null float64     with missing records, dropping it
students_clean = students_data2.drop(columns=['Unnamed: 3'])
```

```python
# checkin the result for columns names
print(students_clean.columns)
```

```python
# info about students data
students_clean.info()
print(students_data.columns)
```

Clearing the “Jobs” table

```python
# info about digital jobs data
jobs_data2.info()
print(jobs_data2.columns)
```

```python
#  checking that the job titles in the Job_Role column are spelled correctly
jobs_data2.info()
print(jobs_data2)
```

```python
# We've found one missing value in the “Vacancies” column. To avoid having to edit it manually in Excel,
# let's search for a solution on Google! We'll fill it in with the column's median and convert it to an integer (int)
jobs_data2['Vacancies'] = jobs_data2['Vacancies'].fillna(jobs_data2['Vacancies'].median()).astype(int)
```

```python
#  checking
jobs_data2.info()
print(jobs_data2)
```

```python
# checkin unique values for salary to see text strings
print(jobs_data2[' Starting Salary'].unique())
```

```python
# there is a hidden space in ' Starting Salary' name, let's fix column names
jobs_data2.columns = ['Job_Role', 'Starting_Salary', 'Vacancies']
```

```python
# removing '£' and ',' symbols from column because they block math analysis
# I had to do a bit of Googling to figure out how to do it
jobs_data2['Starting_Salary'] = jobs_data2['Starting_Salary'].str.replace('£', '', regex=False)
jobs_data2['Starting_Salary'] = jobs_data2['Starting_Salary'].str.replace(',', '', regex=False)
```

```python
# CASTING TO INTEGER to calculate later
jobs_data2['Starting_Salary'] = jobs_data2['Starting_Salary'].astype(int)
```

```python
# checkin the result for cleaned salary column
print(jobs_data2['Starting_Salary'].unique())
```

Cleaning historical course data

```python
# info about historical data
course_history2.info()
print(course_history2.columns)
```

```python
# first column with course names has no name and shows as 'Unnamed: 0'
# replacement for a proper clear name 'Subject'
course_history2.rename(columns={'Unnamed: 0': 'Subject'}, inplace=True)
```

```python
# checking the data structure before saving
course_history2.info()
```

```python
# GOOGLE SEARCH NOTE: We did not cover how to save multiple sheets into one file in our lessons.
# I used Google and StackOverflow to find a solution, and found pd.ExcelWriter function.
# This helps to keep all cleaned college data in one file but on separate tabs for future Power BI use.
with pd.ExcelWriter('Combined_College_Data.xlsx') as writer:
    students_clean.to_excel(writer, sheet_name='Students_Info', index=False)
    jobs_data2.to_excel(writer, sheet_name='Digital_Jobs', index=False)
    course_history2.to_excel(writer, sheet_name='Historical_Course_Data', index=False)

print("Data successfully cleaned and migrated! Well done mate")
```
