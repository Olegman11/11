Python Control Structure

```python
# functions if, and, elif, else
age =17
if age > 0 and age < 12:
  print('You are a child')
elif age >= 13 and age <= 17:
  print('You are a teenager')
elif age >= 18 and age < 69:
  print('You are an Adult')
else: print('you are so old')
```

```python
# function 'and'
a= 200
b= 330
c= 500
if a > b and c > a:
  print('Both conditions are True')
else: print('Faulse')
```

```python
# function 'or'
a= 200
b= 330
c= 500
if a > b or c > a:
  print('One of conditions is True')
else: print('Faulse')
```

```python
# function 'not'
a= 200
b= 330
if not a > b:
  print('a less than b')
else: print('Faulse')
```

```python
# iteration control structures
# for loop
for i in range(2):
  print('iteration',i)
```

```python
FruitBasket = ['Apple', 'Orange', 'Banano','Kiwi', 'Cherry', 'PineApple']
for i in FruitBasket:
  print(i)
```

```python
myString = 'Data Analytics'
for i in myString:
  print(i)
```

```python
#                                     'break' function
for f in FruitBasket:
  print(f)
  if f == 'Kiwi':
    break
```

```python
# iteration control structures
#                                     'break' function
# for loop
for i in range(7):
  print('iteration',i)
  if i == 5:
    break
```

```python
# iteration control structures
#                                     'break' function + range from... until
# for loop
for i in range(2,7):
  print('iteration',i)
  if i == 5:
    break
```

```python
# iteration control structures
#                                     'break' function + step range
# for loop
for x in range(1,21,2):
  print('iteration',x)
  if x == 20:
    break
```

```python
#   the  while loop function

t = 1
while t <= 7:
  print(t)
  if t == 5:
    break
  t = t+1
```
