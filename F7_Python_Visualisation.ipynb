```python
# importing 'matplotlib' library for visualisation
import pandas as pd               # import pandas
import matplotlib.pyplot as plt   # import matplotlib
import numpy as np                # import numpy
```

```python
# example of plotting a line chart
x_axis = np.array([0,10])
y_axis = np.array([0,100])
plt.plot(x_axis,y_axis)
plt.show()
```

```python
# example of plotting a line chart
x_axis = np.array([0,2,4,8,10])
y_axis = np.array([3,5,7,9,15])
plt.plot(x_axis,y_axis)
plt.title('Plot1 - Heart Pulse') # title of the chart
plt.xlabel('Pulse')    # labeling x-axis
plt.ylabel('distance') # labeling y-axis
plt.grid() # making a grid for this plot
plt.show()
```

```python
# Subplot 1
x_axis = np.array([0,2,4,8,10])
y_axis = np.array([3,5,7,9,15])
plt.plot(x_axis,y_axis)
plt.title('Plot1 - Heart Pulse') # title of the chart
plt.xlabel('Pulse')    # labeling x-axis
plt.ylabel('distance') # labeling y-axis
plt.subplot(1,1,1)
plt.grid() # making a grid for this plot
plt.show()

# Subplot 2
x_axis = np.array([1,6,8,10,14])
y_axis = np.array([4,6,9,11,17])
plt.plot(x_axis,y_axis)
plt.title('Plot1 - Heart Pulse') # title of the chart
plt.xlabel('Pulse')    # labeling x-axis
plt.ylabel('distance') # labeling y-axis
plt.subplot(1,1,1)
plt.grid() # making a grid for this plot
plt.show()
```

```python
# Scatter
x_axis = np.array([1,6,8,10,14,18,24,40])
y_axis = np.array([4,6,9,11,17,21,25,36])
plt.scatter(x_axis,y_axis)
plt.title('Plot1 - Heart Pulse') # title of the chart
plt.xlabel('Pulse')    # labeling x-axis
plt.ylabel('distance') # labeling y-axis
plt.subplot(1,1,1)
plt.grid() # making a grid for this plot
plt.show()
```

```python
# bar
x_axis = np.array(['pen','pencil','paper','blades','roll'])
y_axis = np.array([4,14,9,25,17])
plt.bar(x_axis,y_axis)
plt.title('Sales 2026') # title of the chart
plt.xlabel('Goods')    # labeling x-axis
plt.ylabel('Sales number') # labeling y-axis
plt.grid() # making a grid for this plot
plt.show()
```

```python
# histogram

x_axis = np.array([100,14,100])
plt.hist(x_axis)
plt.title('Sales 2026') # title of the chart
plt.xlabel('Goods')    # labeling x-axis
plt.ylabel('Sales number') # labeling y-axis
plt.grid() # making a grid for this plot
plt.show()
```
