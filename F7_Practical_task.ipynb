During the initial data analysis, I identified several data quality issues that need to be addressed before moving to the next stage of analysis.

First, the dataset contains 151 rows and 7 columns, but several columns have missing values. For example, the Age column contains only 127 non-null values, and SystolicBP contains 124 non-null values, while other columns contain 143 non-null values. This suggests that there are missing records, and I need to investigate whether these values were lost during data collection or data import.

Second, I found inconsistent data types. The DiastolicBP column is stored as an object instead of a numeric type. After reviewing the data, I noticed non-numeric values such as "Eighty", which should be converted into a proper numeric format.

Third, I found inconsistent category labels in the RiskLevel column. For example, the same categories appear in different formats, such as "high_risk", "high risk", "LOW", and numeric values like 0 and 1. These values need to be standardised to ensure consistency during analysis.

I also noticed possible outliers or unusual values in some numerical columns. For example, the BS column contains values with a very wide range, such as 0.1 and 14. I need to understand the meaning and unit of this variable before deciding whether these values are valid or represent data entry errors.

```python
#checking RiskLevel for unique values
data['RiskLevel'].unique()
```

```python
# replacement the incorrect data for 'High Risk' and 'Low Risk' columns
data['RiskLevel'] = data['RiskLevel'].replace(['high_risk', 'high risk', '1'], 'High Risk')
data['RiskLevel'] = data['RiskLevel'].replace(['low risk', '0', 'LOW'], 'Low Risk')
```

```python
# drop the row with NaN in RiskLevel column
data.dropna(subset=['RiskLevel'], inplace=True)
```

```python
data
```

```python
#checking HeartRate for unique values
data['HeartRate'].unique()
```

```python
# sorting out with 250 HeartRate replacing for average
mean_HeartRate = data['HeartRate'].mean()
data.replace(250, mean_HeartRate, inplace=True)
```

```python
# formating the data HeartRate
mean_HeartRate = data['HeartRate'].mean()
data['HeartRate'].fillna(mean_HeartRate, inplace=True)
data['HeartRate'] = round(data['HeartRate'], 1)
data['HeartRate'].unique()
```

```python
data['BodyTemp'] = Data_health_risk['BodyTemp']
```

```python
#checking BodyTemp for unique values
data['BodyTemp'].unique()
```

```python
# Anything above 50 is in degrees Fahrenheit. Anything below that is in degrees Celsius. I need Celsius
data.loc[data['BodyTemp'] > 50, 'BodyTemp'] = (data['BodyTemp'] - 32) / 1.8
data['BodyTemp'].unique()
```

```python
# formating the data BodyTemp
mean_BodyTemp = data['BodyTemp'].mean()
data['BodyTemp'].fillna(mean_BodyTemp, inplace=True)
data['BodyTemp'] = round(data['BodyTemp'], 1)
data['BodyTemp'].unique()
```

```python
#checking BS for unique values
data['BS'].unique()
```

```python
# sorting out with 0,1 BS replacing for average
mean_bs = data['BS'].mean()
data.replace(0.1, mean_bs, inplace=True)
```

```python
data['BS'].fillna(mean_bs, inplace=True)
data['BS'] = round(data['BS'], 1) #round characters
data['BS'].unique()
```

```python
#checking DiastolicBP for unique values
data['DiastolicBP'].unique()
```

```python
# replacement the incorrect data
data['DiastolicBP'] = data['DiastolicBP'].replace('Eighty', 80)
data['DiastolicBP'] = data['DiastolicBP'].replace('Seventy', 70)
data['DiastolicBP'] = data['DiastolicBP'].replace('Sixty', 60)
data['DiastolicBP'].unique()
```

```python
#  Removing Unknown
data['DiastolicBP'] = data['DiastolicBP'].replace('Unknown', np.nan)
mean_dia = data['DiastolicBP'].mean()
data['DiastolicBP'].fillna(mean_dia, inplace=True)
```

```python
#  convertin characters (Casting)
data['DiastolicBP'] = data['DiastolicBP'].astype(int)
```

```python
# checkin the result for column DiastolicBP
data['DiastolicBP'].unique()
```

```python
#checking SystolicBP for unique values
data['SystolicBP'].unique()
```

```python
#  Removing nan
mean_sys = data['SystolicBP'].mean()
data['SystolicBP'].fillna(mean_sys, inplace=True)
#  convertin characters to integer (Casting)
data['SystolicBP'] = data['SystolicBP'].astype(int)
data['SystolicBP'] = round(data['SystolicBP'] / 10) * 10 #round to the correct format
data['SystolicBP'] = data['SystolicBP'].astype(int)
data['SystolicBP'].unique()
```

```python
#checking Age for unique values
data['Age'].unique()
```

```python
# sorting out with 150-years old guy
mean_age = data['Age'].mean()
data.replace(150, mean_age, inplace=True)
data['Age'].unique()
```

```python
# formating the data column Age
mean_Age = data['Age'].mean()
data['Age'].fillna(mean_Age, inplace=True)      # and get a mean value (average) from the colum
data['Age'] = data['Age'].astype(int)           # CASTING TO INTEGER
```

```python
#checking Age for unique values
data['Age'].unique()
```
